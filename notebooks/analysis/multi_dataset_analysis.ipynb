```python
winner_counts = pd.crosstab(meta["dataset"], meta["best_tsfm"])
winner_counts
```

```python
winner_rates = winner_counts.div(winner_counts.sum(axis=1), axis=0)
winner_rates.round(4)
```

The above is **RQ1**: 
> How does relative TSFM performance differ across datasets?

```python
tsfm_columns = ["chronos2_mase", "moirai2_mase", "timesfm3_mase",]
fixed_performance = meta[tsfm_columns].mean().sort_values()
fixed_performance
```

```python
best_fixed_column = fixed_performance.index[0]
best_fixed_mase = fixed_performance.iloc[0]

best_fixed_column, best_fixed_mase
```

```python
oracle_mase = meta["oracle_mase"].mean()
oracle_mase
```

```python
routing_opportunity = (best_fixed_mase - oracle_mase) / best_fixed_mase
routing_opportunity
```
